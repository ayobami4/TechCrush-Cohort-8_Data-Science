## 4. Write a program that:
- takes the score of a student (declared)
- Assigns grades:
  - 70-100 : A
  - 60-69 : B
  - 50-59 : C
  - 45-49 : D
  - 0-44 : E
- Prints the grade

```python
# Assign student score
student_score = 99.99

# Assign grades using condional 'if, elif, and else' statements
if student_score >= 70:
  grade = "A"
elif student_score >= 60 and student_score <= 69:
  grade = "B"
elif student_score >= 50 and student_score <= 59:
  grade = "C"
elif student_score >= 45 and student_score <= 49:
  grade = "D"
else:
  grade = "E"

# Print student grade
print(grade)
```

## 5. Create a list of 5 favorite foods:

## 6. Write a program that:
- Uses a loop to calculate the sum of numbers from 1 to 50
- Prints the result.

```python
# Assign a variable
total = 0

# Define a for loop
for number in range(1, 51):
  total += number

# Print the result
print(total)
```
