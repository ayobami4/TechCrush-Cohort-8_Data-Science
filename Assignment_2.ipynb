## 1. Create a NumPy array of 50 random integers between 1 and 100.
- Compute:
  - Mean
  - Median
  - Standard deviation
- Find:
  - Maximum value
  - Minimum value

```python
# Importing the libraries
import numpy as np
import random

# Initializing 50 random integers between 1 and 100 using the for loop
rand = [random.randint(1, 100) for x in range(50)]

# Converting the variable into an array
rand_array = np.array(rand)

# Computing three statistical operations and assign each operation to a varible
mean = np.mean(rand_array)
median = np.median(rand_array)
std = np.std(rand_array, ddof=1)

# Finding the maximum and minimum values
max = np.max(rand_array)
min = np.min(rand_array)

# Printing the results
print(f"Mean: {mean}")
print(f"Median: {median}")
print(f"Standard Deviation: {std}")
print(f"Maximum Value: {max}")
print(f"Minimum Value: {min}")
```

## 2. Using the array created:
- Extract all values:
  - Greater than 60
  - Less than 30

```python
# Extracting values greater than 60
greater_than_60 = rand_array[rand_array > 60]

# Extracting values less than 30
less_than_30 = rand_array[rand_array < 30]

# Printing the results
print(f"Values greater than 60: {greater_than_60}")
print(f"Values less than 30: {less_than_30}")
```

## 3. Using the employee dataset I sent in one of our previous classes:
- Display the first 5 rows
- Display summary statistics
- Identify the job title with the highest salary_year_avg
- Count the number of unique job titles

```python
# Importing the pandas library
import pandas as pd

# Importing the dataset
df = pd.read_csv("/content/employee_dataset.csv")

# Displaying the first 5 rows
first_5_rows = df.head(5)

# Displaying summary statistics
summary_stat = df.describe()

# Identifying the job title with the highest salary year average
highest_salary_year_avg = df["salary_year_avg"].max()
job_title_with_highest_salary_year_avg = df.loc[df["salary_year_avg"].idxmax(), "job_title"]

# Counting the number of unique job titles
number_of_each_unique_value = df["job_title"].value_counts()
total_number_of_all_unique_values = number_of_each_unique_value.count()

# Printing the results
print(f"First 5 rows: {first_5_rows}")
print(f"Summary Statistics: {summary_stat}")
print(f"Job Title with the Highest Salary Year Average is: {job_title_with_highest_salary_year_avg}, with a salary year average of: {highest_salary_year_avg}")
print(f"Number of each Unique Job Titles include: {number_of_each_unique_value}, with a total number of all unique values of: {total_number_of_all_unique_values}")
```

## 4. Using the same dataset:
- Filter `salary_year_avg` above 70000
- Sort job title by `salary_year_avg` (descending order)

```python
# Filtering `salary_year_avg` above 70000
filtered_salary_year_avg = df[df["salary_year_avg"] > 70000]

# Sorting job title by `salary_year_avg` in descending order
sorted_salary_year_avg = df.sort_values(by="salary_year_avg", ascending=False)
sorted_salary_year_avg = sorted_salary_year_avg[["job_title", "salary_year_avg"]]

# Printing the result
print(f"Filtered Salary Year Average above 70000: {filtered_salary_year_avg}")
print(f"Sorted Salary Year Average in descending order by Job Title: {sorted_salary_year_avg}")
```

## 5. Using the same data
- Plot a bar chart of job title vs `salary_year_avg`
- Plot a line graph of `salary_year_avg`
- Add:
  - Title
  - Axis Labels

```python
# Importing the matplotlib library
import matplotlib.pyplot as plt

# Plotting a bar chart of job title vs `salary_year_avg'
plt.figure(figsize=(10, 6))
plt.bar(df["job_title"], df["salary_year_avg"])

# Adding title and axis labels
plt.title("Job Title vs Salary Year Average")
plt.xlabel("Job Title")
plt.ylabel("Salary Year Average")
plt.grid(False)

# Rotating job titles for readability
plt.xticks(rotation=45)

plt.show()

# Ploting a line graph of `salary_year_avg`
plt.plot(df["salary_year_avg"])
plt.title("Salary Year Average")
plt.xlabel("Year")
plt.ylabel("Salary Year Average")
plt.grid(False)
plt.show()
```

# **Bonus Question (Optional)**

- Group job title by `salary_year_avg`
- Visualize using the visual you deem fit.

```python
# Grouping job title by `salary_year_avg"
group = df.groupby("job_title")["salary_year_avg"].mean()

# Generating colors from a color map
colors = plt.colormaps["tab10"](range(len(group)))

# Visualizing the group
plt.figure(figsize=(10, 6))
plt.barh(group.index, group.values, color=colors)
plt.title("Job Title vs Salary Year Average")
plt.xlabel("Salary Year Average")
plt.ylabel("Job Title")
plt.grid(False)

plt.show()
```
